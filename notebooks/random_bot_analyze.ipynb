Random bot analyze

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("../data/game_history.csv")
df.describe()
```

Choice Diagram

```python
choices = df['choice'].value_counts()
print(choices)
plt.bar(choices.index, choices.values, color = ["blue", "red", "green", "yellow"])
plt.xlabel("Choice")
plt.ylabel("Count")
plt.title("Choice Distribution")
plt.show()
```

let's see the debt results of the game

```python
avarage = df.groupby('game_id')['debt'].last().mean()
print(f"Average debt: {avarage}")
biggest = df.groupby('game_id')['debt'].last().max()
print(f"Biggest debt: {biggest}")
smallest = df.groupby('game_id')['debt'].last().min()
print(f"Smallest debt: {smallest}")
plt.hist(df.groupby('game_id')['debt'].last(), bins=10, color='blue', edgecolor='black')
plt.xlabel("Debt")
plt.ylabel("Count")
plt.title("Debt Distribution")
plt.show()
```
